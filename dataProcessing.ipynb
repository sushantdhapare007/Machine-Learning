```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.feature_selection import RFE
from sklearn.datasets import load_iris
from imblearn.over_sampling import SMOTE
from sklearn.linear_model import LogisticRegression
```

```python
data = pd.DataFrame({
    'Age': [22, 25, None, 30],
    'Salary': [40000, None, 50000, 60000]
})

#using this we can replace the NaN value or empty value with Mean,Median And Mode
imputer = SimpleImputer(strategy='mean')
data[['Age','Salary']] = imputer.fit_transform(data[['Age','Salary']])

print(data)
```

```python
data.sum()  #this statement gives you the table contain null values or not
```

```python
iris = load_iris()
```

```python
X =iris.data
y = iris.target
```

```python
#1) filter method which starts from 0 features and add up the important features
selector = SelectKBest(score_func=f_classif,k=2)
```

```python
X_new = selector.fit_transform(X,y)
```

```python
X_new.shape
```

```python
X.shape
```

```python
# 2) RFE (Recursive Feature Elimination) which starts from the all features and then removes the less important or the less relevent features
```

```python
model = LogisticRegression(max_iter=300)
```

```python
rfe = RFE(model, n_features_to_select=2)
```

```python
X_rfe = rfe.fit_transform(X,y)
```

```python
print(rfe.support_)
```

```python
#handling imbalenced data 
#type:- Oversampling 

sush = load_iris()
X = sush.data
y= sush.target

smote = SMOTE()
X_reshap,y_reshape = smote.fit_resample(X,y)

print(X_reshap)
```
