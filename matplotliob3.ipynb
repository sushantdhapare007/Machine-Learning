```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6]
y=[2,4,6,8,10,12]
plt.plot(x,y)
plt.title("line Graph")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
names=["sushant","nishant","Vaishnvi","Vaibhavi","Seema"]
marks = [10,50,80,30,60]
plt.title("BAR CHART")
plt.bar(names,marks)
plt.show()
```

```python
x=[1,2,3,4,5,6]
y=[6,5,4,3,2,1]
plt.scatter(x,y)
plt.show()
```

```python
data=[10,20,20,30,40,40,50]
plt.title("Histogram")
plt.hist(data,  bins = 5)
plt.show()
```

```python
label = ["python","java","c++"]
size=[50,30,20]
plt.title("Histogram")
plt.pie(size , labels = label , autopct="%1.1f%%")
plt.show()
```
